```python
"""
Aqua MODIS L3 Chlorophyll-a (MODISA_L3m_CHL) - download + explore
Region: Pakistan coast / Arabian Sea

Install once:
    pip install earthaccess xarray netCDF4 matplotlib numpy
"""
import glob
import os
import earthaccess
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

# ---------------- SETTINGS (yahan badlo) ----------------
START, END = "2023-01-01", "2023-12-31"
WEST, SOUTH, EAST, NORTH = 60, 20, 72, 27     # Pakistan coast / Arabian Sea
OUT_DIR = "./ocean_data"
# --------------------------------------------------------

# 1) Login + search (sirf monthly 4km files)
earthaccess.login()
results = earthaccess.search_data(
    short_name="MODISA_L3m_CHL",
    temporal=(START, END),
    granule_name="*.MO.*.4km.nc",
)
print(f"{len(results)} files mili")

# 2) Download
earthaccess.download(results, OUT_DIR)

# 3) Open + region subset (latitude in files descending hoti hai: north -> south)
files = sorted(glob.glob(f"{OUT_DIR}/*.nc"))
datasets = []
for f in files:
    ds = xr.open_dataset(f)
    sub = ds["chlor_a"].sel(lat=slice(NORTH, SOUTH), lon=slice(WEST, EAST))
    # file name se mahina nikalo, e.g. AQUA_MODIS.20230101_20230131...
    ym = os.path.basename(f).split(".")[1]      # 20230101_20230131
    date = np.datetime64(f"{ym[:4]}-{ym[4:6]}-01")
    datasets.append(sub.expand_dims(time=[date]))
chl = xr.concat(datasets, dim="time")
print(chl)

# 4) Map: saal ka average
plt.figure(figsize=(9, 6))
chl.mean("time", skipna=True).plot(
    cmap="viridis", norm=LogNorm(vmin=0.05, vmax=20),
    cbar_kwargs={"label": "Chlorophyll-a (mg m$^{-3}$)"},
)
plt.title("Aqua MODIS mean chlorophyll-a, Arabian Sea")
plt.savefig("chl_map.png", dpi=150, bbox_inches="tight")

# 5) Monthly time series (region ka average)
ts = chl.mean(dim=["lat", "lon"], skipna=True)
plt.figure(figsize=(9, 4))
ts.plot(marker="o")
plt.ylabel("Mean chlorophyll-a (mg m$^{-3}$)")
plt.title("Monthly mean chlorophyll-a (region average)")
plt.grid(alpha=0.3)
plt.savefig("chl_timeseries.png", dpi=150, bbox_inches="tight")

plt.show()
```
